```python
# Fit and plot spline to histograms 

best_fit_torch_rcusp = 0.048
observed_torch_rcusp = 0.426

bins = np.linspace(-1, 1, 7)  # Define the bins for the histogram
print("Bins:", bins)

plt.figure(figsize=(11, 8))

bins_fine = np.linspace(bins[0], bins[-2], 20)


# Create a spline for each histogram
spline_150 = UnivariateSpline(bins[:-1], np.histogram(onefifty_pc, bins=bins)[0], k=3, s=0)
spline_240 = UnivariateSpline(bins[:-1], np.histogram(twoforty_pc, bins=bins)[0], k=3, s=0)
spline_60 = UnivariateSpline(bins[:-1], np.histogram(sixty_pc, bins=bins)[0], k=3, s=0)

# Plot the splines
plt.plot(bins_fine, spline_60(bins_fine), label='$2.0 \\times 10^{-22}$ eV', linewidth=3)
plt.plot(bins_fine, spline_150(bins_fine), label='$0.8 \\times 10^{-22}$ eV', linewidth=3)
plt.plot(bins_fine, spline_240(bins_fine), label='$0.5 \\times 10^{-22}$ eV', linewidth=3)
plt.hist(observed_rcusp, bins=np.linspace(-1, 1, 11), alpha=0.5, label='Keeton et al 2003', color='black', edgecolor='white')
plt.axvline(x=best_fit_torch_rcusp, color='black', linestyle='--', label='Smooth Model Best Fit \n $R_{cusp}$ = ' + f'{best_fit_torch_rcusp}', linewidth=3)
plt.axvline(x=observed_torch_rcusp, color='red', linestyle='-.', label='Observed $R_{cusp}$ = ' + f'{observed_torch_rcusp}', linewidth=3)
# plt.hist(rcusp_values, bins=np.linspace(-1, 1, 11), alpha=0.5, label='MCMC', color='orange', edgecolor='white')

# Increase the font size for the ticks
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)

plt.xlim(-1.1, 0.75)

# Y ticks
plt.yticks(np.arange(0, 23, 4))

plt.title('Wave Dark Matter $R_{cusp}$ Distribution', fontsize=20)
plt.xlabel('$R_{cusp}$', fontsize=20)
plt.ylabel('N', fontsize=20)
plt.legend(fontsize=16)
plt.show()
```
